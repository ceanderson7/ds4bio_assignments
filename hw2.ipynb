```python
class mydata:
    def __init__(self, dataset): 
        self.dataset=dataset
    
    def mean(self):
        total = 0 
        for element in self.dataset:
            total = total + element
        return total/len(self.dataset)
    
    def std(self):
        mean_value = self.mean()
        numerator = 0 
        for element in self.dataset:
            numerator = numerator + (element -  mean_value)**2

        return (numerator/len(self.dataset))**0.5



x = mydata([1, 2, 7, 5])    
print(x.mean())
print(x.std())
```
