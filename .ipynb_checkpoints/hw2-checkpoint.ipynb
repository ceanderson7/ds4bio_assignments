```python
def sorting_hat():
    Q1 = input("Does your ambition drive you?")
    Q2 = input("Are you more brave than ambitious")
    if Q1 == "Y" and Q2 == "N":
        house = "Slytherine"
    elif Q1 == "N" and Q2 == "N":
        house = "Slytherine"
    else:
        house = "Gryffindor"

    return house 
    
print(sorting_hat())
  
```

```python
class mydata:
    def __init__(self, dataset): 
        self.dataset=dataset
    
    def mean(self):
        total = 0 
        for element in self.dataset:
            total = total + element
        return total/len(self.dataset)
    
    def std(self):
        mean_value = self.mean()
        numerator = 0 
        for element in self.dataset:
            numerator = numerator + (element -  mean_value)**2

        return (numerator/len(self.dataset))**0.5



x = mydata([1, 2, 7, 5])    
print(x.mean())
print(x.std())
```

```python
import numpy as np
a = np.array([1, 6, 8, 4], dtype=float)
def lag(a, integer):
    if integer > -1 and integer < len(a)+1:
        lagged_data = np.empty(len(a))
        for i in range(len(a)):
            if (i+integer) < len(a):
                lagged_data[i+integer]= a[i]           
        for i in range(integer):
            lagged_data[i] = np.nan
        return lagged_data
    return "integer not in range"
print(lag(a, -1))
print(lag(a, 0))
print(lag(a, 1))
print(lag(a, 2))
print(lag(a, 3))
print(lag(a, 4))
print(lag(a, 5))
```
