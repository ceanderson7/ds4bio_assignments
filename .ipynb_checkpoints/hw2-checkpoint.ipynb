```python
def sorting_hat():
    Q1 = input("Does your ambition drive you?")
    Q2 = input("Are you more brave than ambitious")
    if Q1 == "Y" and Q2 == "N":
        house = "Slytherine"
    elif Q1 == "N" and Q2 == "N":
        house = "Slytherine"
    else:
        house = "Gryffindor"

    return house 
    
print(sorting_hat())
  
```
